```python
import kaggle_benchmarks as kbench

@kbench.task(name="kubernetes deployment manifest for nginx")
def kubernetes_deployment_manifest_for_nginx(llm):
    prompt = """Write a Kubernetes Deployment manifest (apps/v1) for an nginx image.
    Include the following specifications:
    - 3 replicas
    - A readiness probe on path '/healthz' and port 80
    - Resource requests: cpu=100m and memory=128Mi
    - Output ONLY the YAML content without any markdown formatting or explanation."""

    response = llm.prompt(prompt)

    # Use a judge to evaluate the structural and specific requirements of the YAML
    assessment = kbench.assertions.assess_response_with_judge(
        criteria=[
            "The output is a valid Kubernetes Deployment manifest (apiVersion: apps/v1, kind: Deployment).",
            "The manifest specifies 3 replicas.",
            "The container image used is 'nginx'.",
            "The readiness probe is correctly configured for path '/healthz' on port 80.",
            "The resource requests are correctly set to cpu: 100m and memory: 128Mi.",
            "The response contains only YAML, with no markdown code fences or surrounding text."
        ],
        response_text=response,
        judge_llm=kbench.judge_llm
    )

    if assessment is None:
        kbench.assertions.assert_fail(expectation="Judge LLM failed to return a valid assessment.")
        return

    # Convert judge results into formal assertions
    for result in assessment.results:
        kbench.assertions.assert_true(
            result.passed,
            expectation=f"Criterion '{result.criterion}' failed: {result.reason}"
        )

kubernetes_deployment_manifest_for_nginx.run(kbench.llm)
```
